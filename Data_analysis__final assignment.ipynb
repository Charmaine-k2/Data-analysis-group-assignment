```python
#(3a) barplot poverty vs learning poverty

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

avg_poverty = poverty["OBS_VALUE"].mean()
avg_learning_poverty = learning_poverty["OBS_VALUE"].mean()

data = pd.DataFrame({"Category": ["Poverty", "Learning Poverty"],"Average Value": [avg_poverty, avg_learning_poverty]})
plt.figure(figsize=(6, 4))
sns.barplot(data=data, x="Category", y="Average Value", palette="Blues_d")
plt.yscale("log")
plt.title("Average Values (Log Scale): Poverty vs Learning Poverty")
plt.ylabel("Average OBS_VALUE (log scale)")
plt.xlabel("")
plt.tight_layout()
plt.show()
```

# 1.Bar plot [Average Poverty vs Learning Poverty]
From the bar plot, it is clear that the average value for Learning Poverty is much higher than that of Poverty. Since the graph is presented on a log scale, the two bars appear closer in height; however, in reality, the difference between them is very large.

 I can confirm that while poverty affects people on a bigger scale, learning poverty is still a major because it limits education quality for many children.

 The substantial gap between poverty and learning poverty highlights the fact that income improvements alone do not automatically translate into educational gains. This suggests that structural challenges within education systems persist despite economic progress.

Elevated levels of learning poverty can contribute to the intergenerational transmission of income poverty, as children who fail to acquire basic literacy and numeracy skills are more likely to experience limited employment opportunities in adulthood.

The use of a log scale in the visualization underscores the importance of critically evaluating data representation, as statistical techniques can sometimes obscure the magnitude of differences that have significant real-world implications.

```python
#(3b) boxplot distribution comparison

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.DataFrame({"OBS_VALUE": pd.to_numeric(pd.concat([poverty["OBS_VALUE"], learning_poverty["OBS_VALUE"]], ignore_index=True),
        errors="coerce" ),"Category": ["Poverty"] * len(poverty) + ["Learning Poverty"] * len(learning_poverty)})
df.dropna(subset=["OBS_VALUE"], inplace=True)

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Category", y="OBS_VALUE", palette="Blues", showfliers=False)
plt.yscale("log")
plt.title("Distribution Comparison: Poverty vs Learning Poverty (Log Scale, No Outliers)")
plt.ylabel("OBS_VALUE (log scale)")
plt.xlabel("")
plt.tight_layout()
plt.show()
```

# 2. Box Plot [Distribution Comparison]
The box plot shows that Poverty values are more spread out, they vary a lot from country to country, with some being very high and some very low.

This tells me that while poverty levels change a lot depending on the country, learning poverty is consistently a big challenge across many countries, and at higher magnitudes.

The greater variability in poverty levels suggests that some countries have made considerable progress in reducing income poverty, while others still face extreme deprivation.

In contrast, the relatively narrower distribution of learning poverty highlights that education quality challenges are widespread and less dependent on a country’s overall economic status.

This pattern indicates that even in wealthier nations with low income poverty, learning poverty can remain high if educational systems are weak or inequitable.

The box plot further suggests that international development goals should treat learning poverty as a universal issue rather than one limited to low-income countries.

```python
#1.a cleaning dataframes

learning_df.drop_duplicates(inplace=True)
poverty_df.drop_duplicates(inplace=True)

learning_df.fillna(0, inplace=True)
poverty_df.fillna(0, inplace=True)

learning_df.columns = learning_df.columns.str.lower().str.replace(' ', '_')
poverty_df.columns = poverty_df.columns.str.lower().str.replace(' ', '_')
```

```python
#b. filtering relevant indicators

poverty_clean = poverty_df[poverty_df['indicator_label'].str.contains('Poverty headcount', case=False)]
learning_clean = learning_df[learning_df['indicator_label'].str.contains('Population', case=False)]
```

```python
# Aggregate learning poverty by country and year
learning_summary = learning_clean.groupby(['ref_area_label', 'time_period'])['obs_value'].sum().reset_index()
learning_summary.rename(columns={'obs_value': 'learning_population'}, inplace=True)
```

```python
#c. merging datasets
merged_df = pd.merge(poverty_clean, learning_summary, on=['ref_area_label', 'time_period'], how='inner')
```

```python
#d. Rename poverty value for clarity
merged_df.rename(columns={'obs_value': 'poverty_rate'}, inplace=True)
merged_df['poverty_impact_score'] = merged_df['poverty_rate'] * merged_df['learning_population'] / 100


merged_df.sort_values(by='poverty_impact_score', ascending=False, inplace=True)
merged_df['poverty_level'] = pd.cut(merged_df['poverty_rate'],  bins=[0, 10, 30, 60, 100], labels=['Low', 'Moderate', 'High', 'Severe'])

print("Merged DataFrame sample:")
display(merged_df.head())
```
